```python
import os
import zipfile
import pandas as pd
from langdetect import detect_langs, LangDetectException
```

```python
comments_zip = "../data/raw/comments.zip"
entries_zip = "../data/raw/entries1.zip"
likes_zip = "../data/raw/likes.zip"

comments_dir = "../data/raw/comments_extracted"
entries_dir = "../data/raw/entries1_extracted"
likes_dir = "../data/raw/likes_extracted"

os.makedirs(comments_dir, exist_ok=True)
os.makedirs(entries_dir, exist_ok=True)
os.makedirs(likes_dir, exist_ok=True)
```

```python
with zipfile.ZipFile(comments_zip, "r") as z:
    z.extractall(comments_dir)

with zipfile.ZipFile(entries_zip, "r") as z:
    z.extractall(entries_dir)

with zipfile.ZipFile(likes_zip, "r") as z:
    z.extractall(likes_dir)

print("Extraction complete.")
```

```python
print("Comments files:", os.listdir(comments_dir))
print("Entries files:", os.listdir(entries_dir))
print("Likes files:", os.listdir(likes_dir))
```
