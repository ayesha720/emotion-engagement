```python
import pandas as pd
```

```python
emotion_df = pd.read_csv(
    "../data/processed/emotion_results.csv"
)

engagement_df = pd.read_csv(
    "../data/processed/engagement_data.csv"
)

print("Emotion shape:", emotion_df.shape)
print("Engagement shape:", engagement_df.shape)
```

```python
final_df = emotion_df.merge(
    engagement_df,
    on="entry_id",
    how="left"
)

print("Final shape:", final_df.shape)

final_df.head()
```

```python
print(
    final_df[
        ["like_count", "CommentCount"]
    ].isna().sum()
)
```

```python
final_df[
    [
        "comment_clean",
        "Top_Emotion",
        "Top_Emotion_Score",
        "like_count",
        "CommentCount"
    ]
].head()
```

```python
final_df.to_csv(
    "../data/processed/final_emotion_engagement.csv",
    index=False
)

print("Final dataset saved.")
print("Shape:", final_df.shape)
```
